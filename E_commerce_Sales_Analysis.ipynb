## 8. Sales Analysis by Category

```python
category_sales = df.groupby("Category")["Total_Sales"].sum()

category_sales
```

```python
df.groupby("Category")["Quantity"].sum()
```

```python
df.groupby("Category")["Total_Sales"].mean()
```

```python
df.groupby("Category")["Order_ID"].count()
```

```python
category_analysis = df.groupby("Category").agg(
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean"),
    Total_Quantity=("Quantity", "sum"),
    Number_of_Orders=("Order_ID", "count")
)

category_analysis
```

```python
category_analysis.sort_values(
    "Total_Sales",
    ascending=False
)
```

## 9. Sales Analysis by City

```python
city_sales = df.groupby("City")["Total_Sales"].sum()

city_sales.sort_values(ascending=False)
```

```python
city_analysis = df.groupby("City").agg(
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean"),
    Total_Quantity=("Quantity", "sum"),
    Number_of_Orders=("Order_ID", "count")
)

city_analysis.sort_values("Total_Sales", ascending=False)
```

## 10. Sales Analysis by Product

```python
product_sales = df.groupby("Product")["Total_Sales"].sum()

product_sales.sort_values(ascending=False)
```

```python
product_analysis = df.groupby("Product").agg(
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean"),
    Total_Quantity=("Quantity", "sum"),
    Number_of_Orders=("Order_ID", "count")
)

product_analysis.sort_values("Total_Sales", ascending=False)
```

```python
product_analysis.sort_values(
    "Total_Sales",
    ascending=False
).head(10)
```

## 11. Sales Analysis by Payment Method

```python
payment_sales = df.groupby("Payment_Method")["Total_Sales"].sum()

payment_sales.sort_values(ascending=False)
```

```python
payment_analysis = df.groupby("Payment_Method").agg(
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean"),
    Total_Quantity=("Quantity", "sum"),
    Number_of_Orders=("Order_ID", "count")
)

payment_analysis.sort_values("Total_Sales", ascending=False)
```

## 12. Sales Analysis by Region

```python
region_analysis = df.groupby("Region").agg(
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean"),
    Total_Quantity=("Quantity", "sum"),
    Number_of_Orders=("Order_ID", "count")
)

region_analysis.sort_values("Total_Sales", ascending=False)
```
